```python
from pathlib import Path
from astropy.io import fits

FOLDER = Path('../../../Documents/astrophotography/Photometry/yzboo/Light/TG/')

fits_files = sorted(FOLDER.glob('*.fit*'))
for f in fits_files:
    with fits.open(f) as hdul:
        obj = hdul[0].header.get('OBJECT', '<missing>')
    print(f'{f.name}: {obj}')
```

```python
from pathlib import Path
from astropy.io import fits

FOLDER = Path('../../../Documents/astrophotography/Photometry/yzboo/Light/TG/')
NEW_OBJECT = 'yzboo'

fits_files = sorted(FOLDER.glob('*.fit*'))
for f in fits_files:
    with fits.open(f, mode='update') as hdul:
        hdul[0].header['OBJECT'] = NEW_OBJECT
    print(f'{f.name}: OBJECT set to {NEW_OBJECT}')
```
