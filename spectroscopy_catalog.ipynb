```python
# Cell 2 — Imports & config

import json
import re
import pandas as pd
from astroquery.simbad import Simbad
from astropy.coordinates import SkyCoord, get_constellation
import astropy.units as u

OUTPUT_FILE = "mobile/spectroscopy.json"
DSO_FILE = "mobile/dso.json"   # objectId collision check only
VS_FILE = "mobile/vs.json"     # objectId collision check only
```

```python
# Cell 3 — Lookup maps

VALID_SUBTYPES = ["O", "B", "A", "F", "G", "K", "M", "Carbon", "Be", "Wolf-Rayet"]

# Default "what to look for" note per subType; a star's Cell 1 "spectralFeatures" overrides it.
DEFAULT_FEATURES = {
    "O":          "Weak Balmer lines; ionized helium (He II 4686) and He I absorption; very blue continuum",
    "B":          "Strong Balmer absorption plus neutral helium (He I 4026, 4471); blue continuum",
    "A":          "Strong hydrogen Balmer absorption series (Hα, Hβ, Hγ, Hδ...)",
    "F":          "Balmer lines weakening; Ca II H&K strengthening; G band (CH 4300) appearing",
    "G":          "Strong Ca II H&K, G band (4300), Mg b (5175), Na D (5890); weak Balmer lines",
    "K":          "Ca I 4227 and strong metal lines, Mg b, Na D; continuum peaks toward the red",
    "M":          "TiO molecular bands give a sawtooth continuum; very little blue light",
    "Carbon":     "C2 Swan bands and CN absorption; almost no blue light",
    "Be":         "Hα (sometimes Hβ) in emission on top of a B-type absorption spectrum",
    "Wolf-Rayet": "Broad emission lines instead of absorption: He II 4686 plus C III/C IV (WC) or N III/N IV (WN)",
}

# Expected leading character of SIMBAD's sp_type per subType, for a sanity warning only.
# Be is absent on purpose: SIMBAD types don't reliably carry the "e" (e.g. del Sco = B0.3IV).
SPTYPE_PREFIX = {s: s for s in "OBAFGKM"} | {"Carbon": "C", "Wolf-Rayet": "W"}

# astropy's get_constellation misspells a few IAU names; correct them to the standard spelling.
ASTROPY_CONSTELLATION_FIXES = {
    "Ophiucus":         "Ophiuchus",
    "Chamaleon":        "Chamaeleon",
    "Crux ":            "Crux",
    "Pisces Austrinus": "Piscis Austrinus",
}

# Greek letters -> SIMBAD-style abbreviations, for objectId derivation.
GREEK = {
    "α": "alf", "β": "bet", "γ": "gam", "δ": "del", "ε": "eps", "ζ": "zet", "η": "eta",
    "θ": "tet", "ι": "iot", "κ": "kap", "λ": "lam", "μ": "mu",  "ν": "nu",  "ξ": "ksi",
    "ο": "omi", "π": "pi",  "ρ": "rho", "σ": "sig", "τ": "tau", "υ": "ups", "φ": "phi",
    "χ": "chi", "ψ": "psi", "ω": "ome",
    "¹": "1", "²": "2", "³": "3",
}
```

```python
# Cell 4 — Manual data
# objectId -> field values that take precedence over SIMBAD (for gaps or bad values).
# RA in decimal hours, Dec in decimal degrees, same as the output.

MANUAL_DATA = {
    # "someobjectid": {"magnitude": 3.1},
}
```

```python
# Cell 5 — SIMBAD lookup

def object_id_for(display_name):
    """Parenthetical designation if present, else the whole name; Greek spelled out."""
    m = re.search(r"\(([^)]+)\)", display_name)
    name = m.group(1) if m else display_name
    name = "".join(GREEK.get(ch, ch) for ch in name)
    return re.sub(r"[^a-z0-9]", "", name.lower())


_simbad = Simbad()
_simbad.add_votable_fields("sp_type", "V", "B")
_simbad.TIMEOUT = 60

ids = [s["simbadId"] for s in stars]
table = _simbad.query_objects(ids)
if len(table) != len(ids):
    raise ValueError(f"SIMBAD returned {len(table)} rows for {len(ids)} ids")
# user_specified_id comes back space-padded (e.g. "10 Lac   ")
by_id = {str(row["user_specified_id"]).strip(): row for row in table}


def _value(row, col):
    v = row[col]
    if getattr(v, "mask", False) or v is None or (isinstance(v, str) and not v.strip()):
        return None
    return v.item() if hasattr(v, "item") else v


lookup_results = []
for star in stars:
    object_id = object_id_for(star["displayName"])
    row = by_id.get(star["simbadId"])
    ra_deg = _value(row, "ra") if row is not None else None
    dec_deg = _value(row, "dec") if row is not None else None
    v_mag = _value(row, "V") if row is not None else None
    b_mag = _value(row, "B") if row is not None else None

    result = {
        "objectId":         object_id,
        "type":             "star",
        "displayName":      star["displayName"],
        "simbadId":         star["simbadId"],
        "simbadMainId":     _value(row, "main_id") if row is not None else None,
        "ra":               ra_deg / 15.0 if ra_deg is not None else None,
        "dec":              dec_deg,
        "subType":          star["subType"],
        "spectralType":     _value(row, "sp_type") if row is not None else None,
        "magnitude":        round(v_mag, 2) if v_mag is not None else None,
        "colorIndexBV":     round(b_mag - v_mag, 2) if b_mag is not None and v_mag is not None else None,
        "calibrationStar":  bool(star.get("calibrationStar", False)),
        "spectralFeatures": star.get("spectralFeatures", DEFAULT_FEATURES.get(star["subType"])),
    }
    result.update(MANUAL_DATA.get(object_id, {}))
    lookup_results.append(result)

print(f"Resolved {sum(r['ra'] is not None for r in lookup_results)}/{len(lookup_results)} stars")
```

```python
# Cell 6 — Build & validate DataFrame

df = pd.DataFrame(lookup_results)

def constellation_from_coords(row):
    if pd.isna(row["ra"]) or pd.isna(row["dec"]):
        return None
    name = get_constellation(SkyCoord(ra=row["ra"] * 15 * u.degree, dec=row["dec"] * u.degree))
    return ASTROPY_CONSTELLATION_FIXES.get(name, name)

df["constellation"] = df.apply(constellation_from_coords, axis=1)

errors = []

missing_coords = df[df["ra"].isna() | df["dec"].isna()]["objectId"].tolist()
if missing_coords:
    errors.append(f"Missing ra/dec for: {missing_coords}")

bad_subtypes = df[~df["subType"].isin(VALID_SUBTYPES)][["objectId", "subType"]]
if not bad_subtypes.empty:
    errors.append(f"Unknown subType (valid: {VALID_SUBTYPES}):\n{bad_subtypes.to_string(index=False)}")

dupes = df[df.duplicated("objectId")]["objectId"].tolist()
if dupes:
    errors.append(f"Duplicate objectIds: {dupes}")

other_ids = {o["objectId"] for o in json.load(open(DSO_FILE))} | {o["objectId"] for o in json.load(open(VS_FILE))}
collisions = sorted(set(df["objectId"]) & other_ids)
if collisions:
    errors.append(f"objectId already used in {DSO_FILE} or {VS_FILE}: {collisions}")

if errors:
    raise ValueError("Validation failed:\n" + "\n".join(errors))

# Warning only: SIMBAD's spectral type disagrees with the curated subType
for _, r in df.iterrows():
    prefix = SPTYPE_PREFIX.get(r["subType"])
    if prefix and r["spectralType"] and not r["spectralType"].startswith(prefix):
        print(f"WARNING: {r['displayName']} is grouped as {r['subType']} but SIMBAD type is {r['spectralType']}")

print(f"Validation passed. {len(df)} stars ready.")
```

```python
# Cell 6b — Full table view for visual verification
# "season" is notebook-only (not written to JSON): when the star transits around 9 pm.

def season_for(ra_hours):
    if 3 <= ra_hours < 9:   return "Winter"
    if 9 <= ra_hours < 15:  return "Spring"
    if 15 <= ra_hours < 21: return "Summer"
    return "Autumn"

df["season"] = df["ra"].map(season_for)

pd.set_option("display.max_rows", 200)
pd.set_option("display.width", 160)

df[["displayName", "objectId", "subType", "spectralType", "magnitude", "colorIndexBV",
    "calibrationStar", "constellation", "season", "ra", "dec"]]
```

```python
# Cell 6c — Seasonal coverage (stars per subType per season)

(pd.crosstab(pd.Categorical(df["subType"], VALID_SUBTYPES), 
             pd.Categorical(df["season"], ["Autumn", "Winter", "Spring", "Summer"]))
   .rename_axis(index="subType", columns="season"))
```

```python
# Cell 7 — Serialize to spectroscopy.json

output_cols = [
    "objectId",
    "type",
    "displayName",
    "ra",
    "dec",
    "subType",
    "spectralType",
    "constellation",
    "magnitude",
    "colorIndexBV",
    "calibrationStar",
    "spectralFeatures",
]

out_df = df[output_cols].copy()
out_df["ra"] = out_df["ra"].round(7)
out_df["dec"] = out_df["dec"].round(7)

with open(OUTPUT_FILE, "w") as f:
    json.dump(json.loads(out_df.to_json(orient="records")), f, indent=2, ensure_ascii=False)
print(f"Written {len(out_df)} stars to {OUTPUT_FILE}")
```
